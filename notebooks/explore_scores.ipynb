# Exploring the Milestone A scores

This notebook parses three original synthetic MusicXML fixtures, inspects explainable features, and rebuilds the offline HTML demonstration. No OMR, LLM, account, network access, or API key is required. Pitches are analyzed as written.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from music_score_analysis import analyze_score, build_report, compare_scores
```

## Parse and inspect

Encoded facts (such as meter and key signature) remain distinct from computed values (such as counts and estimated key).

```python
sample_paths = sorted((ROOT / 'samples').glob('*.musicxml'))
analyses = [analyze_score(path) for path in sample_paths]
[(item['title'], item['note_count'], item['written_pitch_range']) for item in analyses]
```

## Compare

The comparison normalizes the count of notes shorter than one quarter note by each score's note count. Distinct pitch classes ignore octave. These summaries do not establish genre, influence, or broad musical similarity.

```python
comparison = compare_scores(analyses)
comparison
```

## Build the shareable report

The output is a single HTML document with inline styles and charts, so it opens directly from disk.

```python
build_report(analyses, comparison, ROOT / 'demo' / 'index.html')
ROOT / 'demo' / 'index.html'
```
